```python
#Autobiographical number
def findAutoCount(n):
    if n is None:
        return 0
    for i in range(len(n)):
        if n.count(str(i)) != int(n[i]):
            return 0
    return len(set(n))
n=input()
print(findAutoCount(n))
```

```python
#decimal-binary and toggle
n=int(input())
temp=n
binary=""
while temp>0:
    bit=temp % 2
    binary=str(bit)+binary
    temp = temp//2
toggled=""
for bit in binary:
    if bit=='0':
        toggled=toggled+'1'
    else:
        toggled=toggled+'0'
decimal=0
power=1
for bit in toggled[::-1]:
    decimal=decimal+int(bit)*power
    power=power*2
print(decimal)
```

```python
#corona virus
n=int(input())
l=list(map(int,input().split()))
s=int(input())
result=[]
for num in l:
    binary=bin(num)
    binary=binary[2:]
    binary=binary[:-s]
    if binary=="":
        result.append(0)
    else:
        result.append(int(binary,2))
print(*result)
```

```python
#Phone mapping
s=input()
l1=['']
d={
    "2":['a','b','c'],
    "3":['d','e','f'],
    "4":['g','h','i'],
    "5":['j','k','l'],
    "6":['m','n','o'],
    "7":['p','q','r','s'],
    "8":['t','u','v'],
    "9":['w','x','y','z']
}
for i in s:
    l2=[]
    for j in l1:
        for k in d[i]:
            l2.append(j+k)
    l1=l2
print(l1)
```
