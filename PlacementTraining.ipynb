```python
#Autobiographical number
def findAutoCount(n):
    if n is None:
        return 0
    for i in range(len(n)):
        if n.count(str(i)) != int(n[i]):
            return 0
    return len(set(n))
n=input()
print(findAutoCount(n))
```
